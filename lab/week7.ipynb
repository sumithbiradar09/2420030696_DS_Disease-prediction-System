```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

plt.figure(figsize=(6, 4))

sns.histplot(df['sepal length (cm)'], bins=20, kde=True)

plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

plt.show()
```

```python
#Boxplot
plt.figure(figsize=(6, 4))

sns.boxplot(df['sepal length (cm)'])

plt.title("Boxplot of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

plt.show()
```

```python
species_counts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_counts,labels=iris.target_names,autopct='%1.1f%%')
plt.title("Species Distribution (pie chart)")
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.scatterplot(
    x=df['sepal length (cm)'],
    y=df['sepal width (cm)'],
    hue=df['species']
)

plt.title("Scatterplot: Sepal Length vs Sepal Width")
plt.show()

plt.figure(figsize=(6,4))

plt.plot(df['sepal length (cm)'])

plt.title("Line Chart of Sepal Length")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length (cm)")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Ensure the dataset and dataframe are loaded
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

plt.figure(figsize=(6,4))
sns.barplot(x=df['species'], y=df['sepal length (cm)'])
plt.title("Bar chart: Mean sepal length by species")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:,:4].corr(),annot = True, cmap="coolwarm")
plt.title("HeatMap of feature correlations")
plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    df.iloc[:, :4].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.show()

plt.figure(figsize=(6, 4))

plt.scatter(
    df['sepal length (cm)'],
    df['sepal width (cm)'],
    s=df['petal length (cm)'] * 20,
    alpha=0.5,
    c=df['species']
)

plt.title("Bubble Chart: Sepal vs Petal")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

plt.show()
```

```python
sns.pairplot(df.iloc[:,:4])
plt.suptitle("pair plot of iris",y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.violinplot(
    x=df['species'],
    y=df['sepal length (cm)']
)

plt.title("Violin Plot: Sepal Length by Species")
plt.xticks([0, 1, 2], iris.target_names)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

tips = pd.read_csv('tips.csv')
tips.head()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(tips['total_bill'], bins=20, kde=True)
plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x=tips['total_bill'])
plt.title('Boxplot of Total Bill')
plt.xlabel('Total Bill')
plt.show()
```

```python
day_counts = tips['day'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%')
plt.title('Day Distribution (Pie Chart)')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex')
plt.title('Scatterplot: Total Bill vs Tip')
plt.show()

plt.figure(figsize=(6, 4))
plt.plot(tips['total_bill'])
plt.title('Line Chart of Total Bill')
plt.xlabel('Sample Index')
plt.ylabel('Total Bill')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(data=tips, x='day', y='total_bill', errorbar=None)
plt.title('Bar Chart: Mean Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Mean Total Bill')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
num_cols = ['total_bill', 'tip', 'size', 'price_per_person']
sns.heatmap(tips[num_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Heatmap of Feature Correlations')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(tips['total_bill'], tips['tip'], s=tips['size'] * 30, alpha=0.5, c=tips['size'])
plt.title('Bubble Chart: Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
sns.pairplot(tips[['total_bill', 'tip', 'size', 'price_per_person']])
plt.suptitle('Pair Plot of Tips Numeric Features', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.violinplot(data=tips, x='day', y='total_bill')
plt.title('Violin Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
iris = pd.read_csv('Iris (1).csv')
iris.head()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(iris['SepalLengthCm'], bins=20, kde=True)
plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x=iris['SepalLengthCm'])
plt.title('Boxplot of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.show()
```

```python
species_counts = iris['Species'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%')
plt.title('Species Distribution (Pie Chart)')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=iris, x='SepalLengthCm', y='SepalWidthCm', hue='Species')
plt.title('Scatterplot: Sepal Length vs Sepal Width')
plt.show()

plt.figure(figsize=(6, 4))
plt.plot(iris['SepalLengthCm'])
plt.title('Line Chart of Sepal Length')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(data=iris, x='Species', y='SepalLengthCm', errorbar=None)
plt.title('Bar Chart: Mean Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Mean Sepal Length (cm)')
plt.xticks(rotation=15)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
features = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
sns.heatmap(iris[features].corr(), annot=True, cmap='coolwarm')
plt.title('Heatmap of Feature Correlations')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(iris['SepalLengthCm'], iris['SepalWidthCm'], s=iris['PetalLengthCm'] * 20, alpha=0.5)
plt.title('Bubble Chart: Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()
```

```python
sns.pairplot(iris[features + ['Species']], hue='Species')
plt.suptitle('Pair Plot of Iris Features', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.violinplot(data=iris, x='Species', y='SepalLengthCm')
plt.title('Violin Plot: Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.xticks(rotation=15)
plt.show()
```
